```python
def quad(a,b,c):
    x1=(-b+sqrt(b**2-4*a*c))/(2*a)
    x2=(-b-sqrt(b**2-4*a*c))/(2*a)
    return(x1,x2)
```

```python
a,b,c=2,3,-2
x1,x2=quad(a,b,c)
show(f'The roots of the quadratic are {x1} and {x2}')
```

```python
show((x1.n()), (x2).n())
```

```python
def Heron(a,b,c):
    s=(a+b+c)/2.0
    A=sqrt(s*(s-a)*(s-b)*(s-c))
    return A
```

```python
a,b,c=121,161,191
Area=Heron(a,b,c)
print(f'The area of Triangle is {Area.n()}')
```
