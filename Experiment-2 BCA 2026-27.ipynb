Arrange the graphs of  
i)  $x=\sin 3t,\; y=\sin 4t$ 
ii) $x=t-\sin t,\; y=1-\cos t$
iii) $x=4\cos^3 t,\; y=4\sin^3 t$ 
iv) $x=5\cos t+3\cos \tfrac{5t}{3}, 
y=5\sin t-3\sin \tfrac{5t}{3}$
using a two by two grid.

```python
var('t, x, y')

# 1. Lissajous curve
p1 = parametric_plot(
    [sin(3*t), sin(4*t)],
    (t, -2*pi, 2*pi),
    color="green",
    title="Lissajous Curve"
)

# 2. Cycloid
p2 = parametric_plot(
    [t - sin(t), 1 - cos(t)],
    (t, -2*pi, 2*pi),
    color="red",
    title="Cycloid"
)

# 3. Astroid
p3 = parametric_plot(
    [4*cos(t)^3, 4*sin(t)^3],
    (t, -2*pi, 2*pi),
    color="blue",
    title="Astroid"
)

# 4. Hypotrochoid
p4 = parametric_plot(
    [5*cos(t) + 3*cos(5*t/3),
     5*sin(t) - 3*sin(5*t/3)],
    (t, -2*pi, 2*pi),
    color="purple",
    title="Hypotrochoid"
)

L = [p1, p2, p3, p4]
graphics_array(L, 2, 2)
```

Arrange the graphs of  

i)  $x=cos t,\; y=sin t cos t $
ii)  $x=cos t-\frac{1}{3}cos 3t,\;
y=sin t-\frac{1}{3}sin 3t$
iii) $x^2+y^2=9$
iv) $(x^2+y^2)^2=2(x^2-y^2)$
using a one by four grid.

```python
var('t, x, y')

# Parametric: Lemniscate of Gerono
p1 = parametric_plot(
    [cos(t), sin(t)*cos(t)],
    (t, -2*pi, 2*pi),
    color="green",
    title="Lemniscate (Parametric)"
)

# Parametric: Nephroid
p2 = parametric_plot(
    [cos(t) - cos(3*t)/3,
     sin(t) - sin(3*t)/3],
    (t, -2*pi, 2*pi),
    color="red",
    title="Nephroid"
)

# Implicit: Circle
p3 = implicit_plot(
    x^2 + y^2 == 9,
    (x, -4, 4), (y, -4, 4),
    color="blue",
    title="Circle (Implicit)"
)

# Implicit: Lemniscate of Bernoulli
p4 = implicit_plot(
    (x^2 + y^2)^2 == 2*(x^2 - y^2),
    (x, -2, 2), (y, -2, 2),
    color="purple",
    title="Lemniscate (Implicit)"
)

L = [p1, p2, p3, p4]
graphics_array(L, 1, 4)
```

#### Histogram

1. Draw the Histogram for the data set $\{1, 2, 2, 3, 3, 3, 4, 4, 4, 4, 5, 5, 6\}$

```python
# Discrete data
data = [ 1, 1, 2, 2, 2, 3, 3, 4, 4, 4, 4, 5, 6,6]
hist = histogram(data,bins=6)

hist
```

```python
# Sample continuous data
data1 = [2.1, 2.5, 3.0, 3.2, 4.1, 4.7, 5.3, 5.9, 6.0, 6.8]

# Histogram with specified number of bins
hist = histogram(data1, bins=5)

hist
```

```python
#customized class intervals
# Continuous data
data1 = [ 2.5, 3.0, 4.1, 4.7, 5.3, 6.0, 6.8]

# Custom class intervals (bin edges)
bins = [2, 4, 6, 8]

# Histogram with customized frequency classes
hist = histogram(data1, bins=bins)

hist
```
